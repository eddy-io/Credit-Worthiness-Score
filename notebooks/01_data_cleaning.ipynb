## 1. Load the raw data

```python
# Update this path if yours filename is different once downloaded from Kaggle
df = pd.read_csv('../data/raw/cc_approvals.data', header=None)

print(f"Shape: {df.shape}")
df.head()
```

## 3. Standardize missing values

```python
# Replace '?' placeholders with actual NaN so pandas' missing-value tools work correctly
df = df.replace('?', np.nan)

df.isnull().sum()
```

## 4. Rename columns to be human-readable

Some versions of this dataset ship with generic column names (A1, A2, A3...). If yours does, map them

```python
df.columns = ['gender', 'age', 'debt', 'married', 'bank_customer',
            'education_level', 'ethnicity', 
            'years_employed', 'prior_default', 'employed',
            'credit_score', 'drivers_license', 'citizen', 'zip_code', 'income',
            'approved']

df.columns.tolist()
```

```python
# Binary Approved and Denied and change this in data frame
df['approved'] = df['approved'].map({'+':1, '-':0})
```

## 5. Fix data types

Numeric-looking columns sometimes get read in as `object` dtype because of the missing-value placeholders
we just cleaned up. Convert them properly now.

```python
# Identify columns that SHOULD be numeric but got read in as object dtype
# (common after replacing '?' with NaN)
print(df.dtypes)
```

```python
# Example: force specific columns to numeric - edit column names to match data
numeric_cols = []  # e.g. ['age', 'debt', 'years_employed', 'credit_score', 'income']

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

df.dtypes
```

## 6. Handle missing values

Decide a strategy per column type — don't default to dropping everything, since that can bias your
dataset if missingness isn't random. Document *why* you chose each strategy; this is worth a line or
two in your README methodology section.

```python
# Numeric columns: fill with median (robust to outliers, common baseline choice)
numeric_missing = df.select_dtypes(include=[np.number]).columns[
    df.select_dtypes(include=[np.number]).isnull().any()
]
for col in numeric_missing:
    df[col] = df[col].fillna(df[col].median())

# Categorical columns: fill with mode (most frequent value)
categorical_missing = df.select_dtypes(include=['object']).columns[
    df.select_dtypes(include=['object']).isnull().any()
]
for col in categorical_missing:
    df[col] = df[col].fillna(df[col].mode()[0])

print("Remaining missing values:")
df.isnull().sum().sum()
```
